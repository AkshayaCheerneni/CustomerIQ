```python
# Check the data types of all features

print(df.dtypes)
```

```python
# Define categorical features

categorical_features = [
    "gender",
    "location",
    "subscription_type",
    "payment_method",
    "age_group",
    "tenure_group"
]

print("Categorical features:")
print(categorical_features)
```

```python
# Define numerical features

numerical_features = [
    "age",
    "tenure_months",
    "monthly_spend",
    "total_spend",
    "login_frequency",
    "support_tickets",
    "product_usage",
    "discount_used",
    "average_monthly_spend",
    "usage_per_month",
    "support_ticket_rate"
]

print("Numerical features:")
print(numerical_features)
```

```python
print("Categorical columns present:")
print(set(categorical_features).issubset(df.columns))

print("\nNumerical columns present:")
print(set(numerical_features).issubset(df.columns))
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler

print("Preprocessing tools imported successfully.")
```

```python
numeric_transformer = StandardScaler()

categorical_transformer = OneHotEncoder(handle_unknown="ignore")

print("Numeric transformer:", numeric_transformer)
print("Categorical transformer:", categorical_transformer)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

print("Preprocessor created successfully.")
print(preprocessor)
```

```python
from sklearn.pipeline import Pipeline

preprocessing_pipeline = Pipeline(
    steps=[
        ("preprocessor", preprocessor)
    ]
)

print("Preprocessing pipeline created successfully.")
print(preprocessing_pipeline)
```

```python
predictor_features = numerical_features + categorical_features

print("Total predictor features:", len(predictor_features))
print(predictor_features)
```

```python
y_churn = df["churn"]

print("Churn target created.")
print("Shape:", y_churn.shape)
print("\nChurn value counts:")
print(y_churn.value_counts())
```

```python
y_clv = df["customer_lifetime_value"]

print("CLV target created.")
print("Shape:", y_clv.shape)
print("\nCLV summary:")
print(y_clv.describe())
```

```python
X = df[predictor_features]

print("X shape:", X.shape)
print("Churn target shape:", y_churn.shape)
print("CLV target shape:", y_clv.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train_churn, X_test_churn, y_train_churn, y_test_churn = train_test_split(
    X,
    y_churn,
    test_size=0.2,
    random_state=42,
    stratify=y_churn
)

print("Training features:", X_train_churn.shape)
print("Testing features:", X_test_churn.shape)
print("Training target:", y_train_churn.shape)
print("Testing target:", y_test_churn.shape)
```

```python
print("Training churn distribution:")
print(y_train_churn.value_counts(normalize=True))

print("\nTesting churn distribution:")
print(y_test_churn.value_counts(normalize=True))
```

```python
X_train_churn_processed = preprocessing_pipeline.fit_transform(X_train_churn)

print("Churn training data preprocessing completed.")
print("Processed training shape:", X_train_churn_processed.shape)
```

```python
X_test_churn_processed = preprocessing_pipeline.transform(X_test_churn)

print("Churn test data preprocessing completed.")
print("Processed test shape:", X_test_churn_processed.shape)
```

```python
print("Processed churn training shape:", X_train_churn_processed.shape)
print("Processed churn test shape:", X_test_churn_processed.shape)

print("\nProcessed data type:")
print(type(X_train_churn_processed))
```

```python
X_train_clv, X_test_clv, y_train_clv, y_test_clv = train_test_split(
    X,
    y_clv,
    test_size=0.2,
    random_state=42
)

print("Training features:", X_train_clv.shape)
print("Testing features:", X_test_clv.shape)
print("Training target:", y_train_clv.shape)
print("Testing target:", y_test_clv.shape)
```

```python
X_train_clv_processed = preprocessing_pipeline.fit_transform(X_train_clv)

print("CLV training data preprocessing completed.")
print("Processed training shape:", X_train_clv_processed.shape)
```

```python
X_test_clv_processed = preprocessing_pipeline.transform(X_test_clv)

print("CLV test data preprocessing completed.")
print("Processed test shape:", X_test_clv_processed.shape)
```

```python
print("Processed CLV training shape:", X_train_clv_processed.shape)
print("Processed CLV test shape:", X_test_clv_processed.shape)

print("\nProcessed data type:")
print(type(X_train_clv_processed))
```
