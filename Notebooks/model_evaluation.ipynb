```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
classification_results = pd.read_csv("../data/processed/classification_results.csv")
classification_results
```

```python
regression_results = pd.read_csv("../data/processed/regression_results.csv")
regression_results
```

```python
print(classification_results)
```

```python
print(regression_results)
```

```python
best_churn_model = classification_results.loc[
    classification_results["F1 Score"].idxmax()
]

best_churn_model
```

```python
best_clv_model = regression_results.loc[
    regression_results["R2"].idxmax()
]

best_clv_model
```

```python
print("Best Churn Model:")
print(best_churn_model)

print()

print("Best CLV Model:")
print(best_clv_model)
```

```python
final_summary = pd.DataFrame({
    "Task": ["Churn Prediction", "CLV Prediction"],
    "Best Model": [
        best_churn_model["Model"],
        best_clv_model["Model"]
    ],
    "Main Metric": ["F1 Score", "R2"],
    "Score": [
        best_churn_model["F1 Score"],
        best_clv_model["R2"]
    ]
})

final_summary
```

```python
final_summary.to_csv("../data/processed/final_model_summary.csv", index=False)
```

```python
final_model_summary = pd.read_csv("../data/processed/final_model_summary.csv")
final_model_summary
```

```python
classification_results.plot(
    x="Model",
    y="F1 Score",
    kind="bar",
    legend=False
)

plt.ylabel("F1 Score")
plt.title("Classification Model Comparison")
plt.xticks(rotation=0)
plt.show()
```

```python
regression_results.plot(
    x="Model",
    y="R2",
    kind="bar",
    legend=False
)

plt.ylabel("R2")
plt.title("Regression Model Comparison")
plt.xticks(rotation=0)
plt.show()
```
