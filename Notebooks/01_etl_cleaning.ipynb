```python
import pandas as pd
import numpy as np
from pathlib import Path
```

```python
RAW_DATA_PATH = Path("../data/raw/customerIQ_customer_churn_dataset.csv")
PROCESSED_DATA_PATH = Path("../data/processed/customerIQ_customer_churn_cleaned.csv")

print("Raw data path:", RAW_DATA_PATH)
print("Processed data path:", PROCESSED_DATA_PATH)
```

```python
df = pd.read_csv(RAW_DATA_PATH)

print("Dataset loaded successfully.")
print("Dataset shape:", df.shape)
```

```python
df.head()
```

```python
print(df.columns.tolist())
```

```python
expected_columns = [
    "customer_id",
    "age",
    "gender",
    "location",
    "tenure_months",
    "monthly_spend",
    "total_spend",
    "login_frequency",
    "support_tickets",
    "product_usage",
    "subscription_type",
    "payment_method",
    "discount_used",
    "churn",
    "customer_lifetime_value"
]

missing_columns = set(expected_columns) - set(df.columns)
extra_columns = set(df.columns) - set(expected_columns)

print("Missing expected columns:", missing_columns)
print("Unexpected columns:", extra_columns)
```

```python
print(df.dtypes)
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values)
```

```python
duplicate_rows = df.duplicated().sum()

print("Duplicate rows:", duplicate_rows)
```

```python
duplicate_customer_ids = df["customer_id"].duplicated().sum()

print("Duplicate customer IDs:", duplicate_customer_ids)
```

```python
duplicate_customer_records = df[
    df["customer_id"].duplicated(keep=False)
]

print(duplicate_customer_records)
```

```python
numeric_columns = [
    "age",
    "tenure_months",
    "monthly_spend",
    "total_spend",
    "login_frequency",
    "support_tickets",
    "product_usage",
    "churn",
    "customer_lifetime_value"
]

df[numeric_columns].describe()
```

```python
negative_values = (df[numeric_columns] < 0).sum()

print("Negative values:")
print(negative_values)
```

```python
rows_before = len(df)

df = df.drop_duplicates()

rows_after = len(df)

print("Rows before:", rows_before)
print("Rows after:", rows_after)
print("Rows removed:", rows_before - rows_after)
```

```python
missing_after_cleaning = df.isnull().sum()

print("Missing values after duplicate removal:")
print(missing_after_cleaning)
```

```python
categorical_columns = [
    "gender",
    "location",
    "subscription_type",
    "payment_method"
]

for column in categorical_columns:
    df[column] = df[column].astype("string").str.strip()
```

```python
print(df["discount_used"].value_counts(dropna=False))
```

```python
print("Final dataset shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nMissing values:")
print(df.isnull().sum())
```

```python
PROCESSED_DATA_PATH.parent.mkdir(
    parents=True,
    exist_ok=True
)

print("Processed folder is ready.")
```

```python
df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(
    f"Cleaned dataset saved to: "
    f"{PROCESSED_DATA_PATH}"
)
```

```python
cleaned_df = pd.read_csv(PROCESSED_DATA_PATH)

print("Processed dataset loaded successfully.")
print("Shape:", cleaned_df.shape)

cleaned_df.head()
```

```python
missing_values = df.isnull().sum()

print("Missing values in each column:")
print(missing_values[missing_values > 0])
```

```python
numeric_missing_columns = [
    "age",
    "monthly_spend",
    "login_frequency",
    "product_usage"
]

print("Median values:")
print(df[numeric_missing_columns].median())
```

```python
for column in numeric_missing_columns:
    df[column] = df[column].fillna(df[column].median())

print("Missing numerical values have been filled.")
```

```python
remaining_missing_values = df.isnull().sum()

print("Remaining missing values:")
print(remaining_missing_values[remaining_missing_values > 0])
```

```python
total_missing_values = df.isnull().sum().sum()

print("Total missing values:", total_missing_values)
```

```python
duplicate_rows_after_cleaning = df.duplicated().sum()

print("Duplicate rows after cleaning:", duplicate_rows_after_cleaning)
```

```python
duplicate_customer_ids_after_cleaning = df["customer_id"].duplicated().sum()

print(
    "Duplicate customer IDs after cleaning:",
    duplicate_customer_ids_after_cleaning
)
```

```python
negative_values_after_cleaning = (
    df[numeric_columns] < 0
).sum()

print("Negative values after cleaning:")
print(negative_values_after_cleaning)
```

```python
print("Final cleaned dataset shape:", df.shape)
```

```python
print("===== FINAL DATA QUALITY REPORT =====")

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate customer IDs:", df["customer_id"].duplicated().sum())
```

```python
df.to_csv(
    PROCESSED_DATA_PATH,
    index=False
)

print(f"Updated cleaned dataset saved to: {PROCESSED_DATA_PATH}")
```
